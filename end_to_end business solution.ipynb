## 6.Transformation

```python
sales=order_items[["order_id","product_id","price","freight_value"]].merge(
    orders[["order_id","customer_id","order_purchase_timestamp","order_status"]],
    on="order_id")
```

```python
sales.shape
```

```python
sales=sales.merge(
    products[["product_id","product_category_name"]],
    on="product_id"
)
```

```python
sales.shape
```

```python
sales["sales_amount"]=sales["price"]+sales["freight_value"]
```

```python
sales.shape
```

## Date transformation

```python
sales["order_purchase_timestamp"]=pd.to_datetime(sales["order_purchase_timestamp"])
```

```python
sales["year"]=sales["order_purchase_timestamp"].dt.year
sales["month"]=sales["order_purchase_timestamp"].dt.month
```

```python
sales.shape
```
